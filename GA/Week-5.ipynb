# Week 5 GA

Dataset: `Resources/Data/Week-5-GA-Dataset.csv`

Common setup used for all questions:
- The last column (`Credit_Limit`) is the target variable.
- `X` contains all feature columns and `y` contains the target column.
- Train-test split uses a 70:30 ratio with `random_state=42`.

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score, root_mean_squared_error
```

```python
df = pd.read_csv('../Resources/Data/Week-5-GA-Dataset.csv')

X = df.iloc[:, :-1]
y = df.iloc[:, -1]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

df.shape
```

## Question 1 and 2

Train a `LinearRegression` model with `fit_intercept=False`, compute the test `r2_score`, and find the index of the feature with the highest absolute coefficient.

```python
linear_model = LinearRegression(fit_intercept=False)
linear_model.fit(X_train, y_train)

linear_pred = linear_model.predict(X_test)
q1_r2 = r2_score(y_test, linear_pred)
q2_index = int(abs(linear_model.coef_).argmax())

print(f'Q1 r2_score: {q1_r2:.3f}')
print(f'Q2 highest absolute coefficient index: {q2_index}')
```

## Question 3 and 4

Train a `Ridge` model with:
- `solver='sag'`
- `tol=0.0005`
- `random_state=42`

Compute test `r2_score` and the model intercept.

```python
ridge_model = Ridge(solver='sag', tol=0.0005, random_state=42)
ridge_model.fit(X_train, y_train)

ridge_pred = ridge_model.predict(X_test)
q3_r2 = r2_score(y_test, ridge_pred)
q4_intercept = ridge_model.intercept_

print(f'Q3 r2_score: {q3_r2:.3f}')
print(f'Q4 intercept_: {q4_intercept:.3f}')
```

## Question 5 and 6

Train a `Lasso` model with:
- `alpha=100`
- `random_state=42`

Compute test `r2_score` and count how many coefficients lie in the range `[-1, 1]`.

```python
lasso_model = Lasso(alpha=100, random_state=42)
lasso_model.fit(X_train, y_train)

lasso_pred = lasso_model.predict(X_test)
q5_r2 = r2_score(y_test, lasso_pred)
q6_count = int(((lasso_model.coef_ >= -1) & (lasso_model.coef_ <= 1)).sum())

print(f'Q5 r2_score: {q5_r2:.3f}')
print(f'Q6 coefficients in [-1, 1]: {q6_count}')
```

## Question 7

Train a `KNeighborsRegressor` with:
- `n_neighbors=10`
- `p=1`

Predict on the test set and compute RMSE.

```python
knn_model = KNeighborsRegressor(n_neighbors=10, p=1)
knn_model.fit(X_train, y_train)

knn_pred = knn_model.predict(X_test)
q7_rmse = root_mean_squared_error(y_test, knn_pred)

print(f'Q7 RMSE: {q7_rmse:.3f}')
```

## Question 8

Train a `DecisionTreeRegressor` with:
- `max_depth=10`
- `min_samples_split=6`
- `min_samples_leaf=6`
- `random_state=42`

Predict on the test set and compute RMSE.

```python
tree_model = DecisionTreeRegressor(
    max_depth=10,
    min_samples_split=6,
    min_samples_leaf=6,
    random_state=42,
)
tree_model.fit(X_train, y_train)

tree_pred = tree_model.predict(X_test)
q8_rmse = root_mean_squared_error(y_test, tree_pred)

print(f'Q8 RMSE: {q8_rmse:.3f}')
```

## Question 9 and 10

Perform `GridSearchCV` on `AdaBoostRegressor` with:
- `random_state=42`
- `cv=4`
- `n_estimators=[10, 50, 100, 200, 500]`
- `learning_rate=[0.1, 0.5, 1, 2]`

Use the best model to compute the test `r2_score` and report the best `n_estimators`.

```python
param_grid = {
    'n_estimators': [10, 50, 100, 200, 500],
    'learning_rate': [0.1, 0.5, 1, 2],
}

grid_search = GridSearchCV(
    AdaBoostRegressor(random_state=42),
    param_grid=param_grid,
    cv=4,
)
grid_search.fit(X_train, y_train)

best_model = grid_search.best_estimator_
q9_r2 = best_model.score(X_test, y_test)
q10_n_estimators = grid_search.best_params_['n_estimators']

print(f'Q9 r2_score: {q9_r2:.3f}')
print(f'Q10 best n_estimators: {q10_n_estimators}')
print(grid_search.best_params_)
```
