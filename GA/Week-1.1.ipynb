```python
import pandas as pd

df = pd.read_csv("../Resources/Data/Week-1-GA-Dataset.csv")

print(df.to_string())
```

```python
# Question 1 : Get the number of missing values in the dataframe represented by '?'

print((df == "?").sum())

# NOTE : Remove all the '?' values from the dataframe and replace them with NaN values so that we can easily handle them in the future
df = df.fillna(pd.NA)
print(df.to_string())
```

```python
# Question 2 : Shape of the dataset

print(df.shape)
```

```python
# Question 3 : What data is present at the 692th indexed row and 0th indexed column?

print(df.iloc[692, 0])
```

```python
# Question 4 : What is the value present at the 546th indexed row and 7th indexed column in the data ?

print(df.iloc[546, 7])
```

```python
# Question 5 : What are the unique values present in the Locality feature of the dataset?

print(df["Locality"].unique())
```

```python
# Question 6 : Which of the features have missing (NaN) values present in the dataset?

print(df.isna().any())
```

```python
# Question 7 : Which of the following feature has most missing(NaN) values present in the dataset?

print(df.isna().sum())
```

```python
# Question 8 : Drop all the samples(rows) with missing values strictly greater than 2. How many samples remains after that ?

df = df.dropna(thresh=len(df.columns) - 2)

# for example, if a row has 10 columns and we want to drop rows with more than 2 missing values, we set thresh to 8 (10 - 2) because we want to keep rows that have at least 8 non-missing values

print(len(df))
```

```python
# Question 9 : Drop all the samples(rows) with missing values in the original dataframe. How many samples remains after that ?

df = df.dropna()
print(len(df))
```
