Seasonal-Trend-LOESS(STL) Decomposition

- Variety seasonal pattern
- Handle outliers in time series data
- Only handle additive data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Using airline passengers dataset (monthly totals)
url = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv'
df = pd.read_csv(url, parse_dates=['Month'], index_col='Month' )
df
```

```python
df["Passengers"].plot(figsize=(12,8))
```

```python
from statsmodels.tsa.seasonal import STL
```

```python
res = STL(df["Passengers"]).fit()

res.plot()
plt.figure(figsize=(12,5))
plt.show()
```
