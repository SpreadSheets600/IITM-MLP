## Dataframes

One of the most important data structures in pandas is the DataFrame. A DataFrame is a 2-dimensional labeled data structure with columns of potentially different types. It is similar to a spreadsheet or SQL table, and it is one of the most commonly used data structures in pandas.

Dataframe is like a relational data table where there is a relationship between rows and columns. It is a 2-dimensional data structure that can hold data of different types (integer, float, string, etc.) in columns. Each column can be thought of as a Series, which is a one-dimensional array-like object.

Dataframe can be thought as a matrix with rows and columns, where each column can have a different data type.

````python
# Creating a DataFrame with columns 'name' and 'age'

df = pd.DataFrame({"name": ["Soham", "Samar", "Yash"], "age": [19, 20, 24]})

# Visualizing the dataframe would look like :
#     name   age
# 0  Soham   19
# 1  Samar   20
# 2   Yash   24

# Each column is a Series (1D array with index).
# Rows are indexed starting from 0 by default.

type(df)

# To access a column, we can use the column name as an attribute or as a key

df.name  # Accessing 'name' column as an attribute
df["name"]  # Accessing 'name' column as a key

# To access a particular row, we can use the .loc or .iloc indexers. .loc is label-based, while .iloc is integer position-based.

print(df.loc[0])  # Accessing the first row using .loc
print(df.iloc[0])  # Accessing the first row using .iloc

# accessing row with a element
print(df.loc[df["name"] == "Samar"])  # Accessing row where name is 'Samar'

# updating a value in the dataframe, for example changing the age of Samar to 21
df.loc[df["name"] == "Samar", "age"] = 21

# we can add new columns, wehen doing so each and every row must have the data

df["job"] = ["Nalla", "Intern", "Job"]

# we can also add a new row and for that we need to give a dictionary with the given key value pairs

new_row = pd.DataFrame({"name": ["No Idea"], "age": ["No Idea"], "job": ["No Idea"]})

# now we will con-catinate the old df and new df using concat function and passing old and new df in a list as a param, we can essentailly add n number of rows

df = pd.concat([df, new_row])
print(df)

# if while concatinating, a coloumn has a data which the other df dosen't it will default as NaN -
# for example if df_1 has column phone, and df_2 while concatinating dosen't have phone it would be NaN by deafult
````

## Series

A Series is a one-dimensional labeled array that can hold data of any type (integer, float, string, etc.). It is similar to a column in a DataFrame. Each element in a Series has an associated label, which is called the index. The index can be thought of as the row labels in a DataFrame.

A Series can be created from a list, a NumPy array, or a dictionary. It can also be created from a DataFrame by selecting a single column.

A Series can be thought of as a single column of a DataFrame, and it can be used to perform various operations such as indexing, slicing, and mathematical operations. Or it can be thought of a vector of data with an associated index.

````python
# Creating a series with columns 'name'

sr = pd.Series(["Soham", "Samar", "Yash"])

# Visualizing the series would look like :
# 0 Soham
# 1 Samar
# 2 Yash

# Each element is a element of the 1D array.
# Each element is indexed starting from 0 by deafult.

type(sr)
print(sr)

# we can have a custom index as well, by passing the keyword index

sr = pd.Series(
    ["Soham", "Samar", "Yash"], index=["a", "b", "c"]
)  # It is considered as a lable, and can contain anything

print(sr)

# To access an element directly by the lable, we use the loc operator
# Accessing Samar using the lable we will do -

print(sr.loc["b"])  # Return the value where lable is b

# Using loc we can update the elements easily

sr.loc["c"] = "Yashuuu"
print(sr)

# if I have to return using the integer index value then we will use the iloc operator

print(sr.iloc[1])  # will return samar cause according to numerical index Samar is at 1

# NOTE : index must always match the number of elements, if that's not the case then it will turn a ValueError

# We can easily perform filter by providing the condition in []
# for example i am trying to get the name of the user who has letter u in his name

print(sr[sr.str.contains("u")])

# here i am using .str.contains() which checks each element of series for a substring presence and returns a boolean

# We can use dic as a series as well -
calories = {"Day 1": 1750, "Day 2": 100, "Day 3": 700}

sr_2 = pd.Series(calories)
sr_2.loc["Day 2"] += 500 # adding 500 more to day 2

print(sr_2)
````

## Dataframe Attributes

1. `shape` : This attribute returns the number of rows and columns in the DataFrame as a tuple (rows, columns).
2. `dtypes` : This attribute returns the data types of each column in the DataFrame as a series.
---
3. `info()` : This method provides a concise summary of the DataFrame, including the number of non-null values, data types, and memory usage.
4. `describe()` : This method generates descriptive statistics of the DataFrame, such as count, mean, standard deviation, minimum, and maximum values for each numeric column.
---
5. `head(n)` : This method returns the first n rows of the DataFrame. By default, it returns the first 5 rows.
6. `tail(n)` : This method returns the last n rows of the DataFrame. By default, it returns the last 5 rows.
7. `sample(n)` : This method returns a random sample of n rows from the DataFrame. By default, it returns 1 row.

## Basic Statistics



````python
# aggregate functions are used to perform some operations on the data and return a single value

# mean aggregate function is used to get the mean of all the values in a column, for example if we want to get the mean of the attack column we can use the mean() function

print(df["Attack"].mean())  # it will return the mean of the attack column
print(
    df.mean(numeric_only=True)
)  # it will return the mean of all the numeric columns in the dataframe

# sum aggregate function is used to get the sum of all the values in a column, for example if we want to get the sum of the attack column we can use the sum() function

print(df["Attack"].sum())  # it will return the sum of the attack column
print(
    df.sum(numeric_only=True)
)  # it will return the sum of all the numeric columns in the dataframe

# min aggregate function is used to get the minimum value in a column, for example if we want to get the minimum value of the attack column we can use the min() function

print(df["Attack"].min())  # it will return the minimum value of the attack column
print(
    df.min(numeric_only=True)
)  # it will return the minimum value of all the numeric columns in the dataframe

# max aggregate function is used to get the maximum value in a column, for example if we want to get the maximum value of the attack column we can use the max() function

print(df["Attack"].max())  # it will return the maximum value of the attack column
print(
    df.max(numeric_only=True)
)  # it will return the maximum value of all the numeric columns in the dataframe

# count aggregate function is used to get the count of all the values in a column, for example if we want to get the count of the attack column we can use the count() function

print(df["Attack"].count())  # it will return the count of the attack column
print(
    df.count(numeric_only=True)
)  # it will return the count of all the numeric columns in the dataframe
````

````python
# grouping is used to group the data based on a column and perform some aggregate function on it, for example if we want to group the data based on the type of pokemon and get the mean of the attack column for each type we can use the groupby() function

grouped = df.groupby("Type 1")  # grouping the data based on the 'Type 1' column
print(grouped["Attack"].mean())  # getting the mean of the attack column for each type
````

````python
# data cleaning - the process of removing or correcting the data which is incorrect, incomplete, irrelevant or duplicated in the dataset

# drop() function is used to drop the rows or columns from the dataframe, for example if we want to drop the column 'Type 2' from the dataframe we can use the drop() function

df = df.drop(
    columns=["Legendary"], errors='ignore'
)  # dropping the 'Legendary' column from the dataframe - since it returns a new dataframe we need to assign it back to df to modify the original dataframe

print(df.to_string())
````

````python
# handing missing data

# check for missing values in the dataframe and if there are any missing values then we can either drop the rows or columns which have missing values or we can fill the missing values with some value, for example if we want to fill the missing values with 0 we can use the fillna() function

df = df.dropna(
    subset=["Type 2"]
)  # dropping the rows which have missing values in the 'Type 2' column

# we will use fillna() function to fill the missing values, NaN with 0 or any other value
df = df.fillna(value=0)  # this will fill all the NaN values in the dataframe with 0

# wse can also pass in dictionary for doing specific column filling
df = df.fillna(
    {"Type 2": "No Type"}
)  # this will fill the NaN values in the 'Type 2' column with 'No Type'

# fixing in-consistent values - for example if we want to replace the value 'Fire' with 'Flame' in the 'Type 1' column we can use the replace() function

df["Type 1"] = df["Type 1"].replace(
    "Fire", "Flame"
)  # replacing 'Fire' with 'Flame' in the 'Type 1' column

# Standardizing the data - for example if we want to standardize the attack column we can use the following formula

df["Attack"] = (df["Attack"] - df["Attack"].mean()) / df["Attack"].std()

# or let's say we want all the Type 1 to be in uppercase we can use the str.upper() function

df["Type 1"] = df[
    "Type 1"
].str.upper()  # converting all the values in the 'Type 1' column to uppercase

# fixing the data types - for example if we want to convert the attack column to integer we can use the astype() function

df["Attack"] = df["Attack"].astype(int)  # converting the 'Attack' column to integer

# removing duplicates - for example if we want to remove the duplicate rows from the dataframe we can use the drop_duplicates() function

df = df.drop_duplicates()  # removing duplicate rows from the dataframe

print(df.to_string())
````
