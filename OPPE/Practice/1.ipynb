## Prepare a data preprocessing pipeline to process features in following order:

1. `Gender`: Impute with most frequent then ordinally encode.
2. `Age`: Impute with mean then standard scale.
3. `HasTension`: Ordinally encode.
4. `AnyHeartDisease`:Ordinally encode.
5. `NeverMarried`:Ordinally encode.
6. `Occupation`: One hot encode.
7. `LivesIn`: Impute with most frequent then ordinally encode.
8. `GlucoseLevel`: Impute with mean, then min-max scaling.
9. `BMI`: Impute with mean, then standard scale.
10. `SmokingStatus`: Impute with most frequent, then one hot encode.

Hint: After transformation, your feature matrix must have columns in following order:

0. Gender
1. Age
2. HasTension
3. AnyHeartDisease
4. NeverMarried
5. Occupation_Govt_job
6. Occupation_Never_worked
7. Occupation_Private
8. Occupation_Self-employed
9. Occupation_children
10. LivesIn
11. GlucoseLevel
12. BMI
13. SmokingStatus_formerly smoked
14. SmokingStatus_never smoked
15. SmokingStatus_smokes

NOTE:
1. Make sure to preprocess the features in the above order exactly. Answer(s) of later question(s) depend(s) upon correct order of featuring processing.
2. You may have to use multiple instances of a trasnformer for this question.

```python
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, StandardScaler, MinMaxScaler

# Make a copy of X_train and X_test to preprocess
X_train_processed = X_train.copy()
X_test_processed = X_test.copy()

# Step 1: Replace implausible values with np.nan in both train and test sets
# Gender: 'Unknown' (as identified in Q2.1)
X_train_processed['Gender'] = X_train_processed['Gender'].replace('Unknown', np.nan)
X_test_processed['Gender'] = X_test_processed['Gender'].replace('Unknown', np.nan)

# Age: <= 0 (as identified in Q2.2)
X_train_processed['Age'] = X_train_processed['Age'].apply(lambda x: np.nan if x <= 0 else x)
X_test_processed['Age'] = X_test_processed['Age'].apply(lambda x: np.nan if x <= 0 else x)

# GlucoseLevel: <= 0 (as identified in Q2.3)
X_train_processed['GlucoseLevel'] = X_train_processed['GlucoseLevel'].apply(lambda x: np.nan if x <= 0 else x)
X_test_processed['GlucoseLevel'] = X_test_processed['GlucoseLevel'].apply(lambda x: np.nan if x <= 0 else x)

# LivesIn: 'Unknown' (as identified in Q2.4)
X_train_processed['LivesIn'] = X_train_processed['LivesIn'].replace('Unknown', np.nan)
X_test_processed['LivesIn'] = X_test_processed['LivesIn'].replace('Unknown', np.nan)

# SmokingStatus: 'Unknown' (as identified in Q2.6)
X_train_processed['SmokingStatus'] = X_train_processed['SmokingStatus'].replace('Unknown', np.nan)
X_test_processed['SmokingStatus'] = X_test_processed['SmokingStatus'].replace('Unknown', np.nan)

# Define preprocessing pipelines for different column types

# 1. Gender: Impute most frequent, then Ordinal Encode
# Assuming default alphabetical order for Male/Female from OrdinalEncoder is acceptable as no specific mapping is given.
gender_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ordinal', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# 2. Age: Impute mean, then Standard Scale
age_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

# 3. HasTension: Ordinal Encode (No=0, Yes=1)
tension_pipeline = Pipeline(steps=[
    ('ordinal', OrdinalEncoder(categories=[['No', 'Yes']], handle_unknown='use_encoded_value', unknown_value=-1))
])

# 4. AnyHeartDisease: Ordinal Encode (No=0, Yes=1)
heartdisease_pipeline = Pipeline(steps=[
    ('ordinal', OrdinalEncoder(categories=[['No', 'Yes']], handle_unknown='use_encoded_value', unknown_value=-1))
])

# 5. NeverMarried: Ordinal Encode (No=0, Yes=1)
nevermarried_pipeline = Pipeline(steps=[
    ('ordinal', OrdinalEncoder(categories=[['No', 'Yes']], handle_unknown='use_encoded_value', unknown_value=-1))
])

# 6. Occupation: One-Hot Encode
# OneHotEncoder will sort categories alphabetically by default, matching the hint's specified order.
occupation_pipeline = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# 7. LivesIn: Impute most frequent, then Ordinal Encode (City=0, Village=1)
livesin_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('ordinal', OrdinalEncoder(categories=[['City', 'Village']], handle_unknown='use_encoded_value', unknown_value=-1))
])

# 8. GlucoseLevel: Impute mean, then Min-Max Scale
glucose_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', MinMaxScaler())
])

# 9. BMI: Impute mean, then Standard Scale
bmi_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

# 10. SmokingStatus: Impute most frequent, then One-Hot Encode
# OneHotEncoder will sort categories alphabetically by default, matching the hint's specified order.
smokingstatus_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# ColumnTransformer to apply all transformations in the specified order
preprocessor = ColumnTransformer(
    transformers=[
        ('gender_proc', gender_pipeline, ['Gender']),
        ('age_proc', age_pipeline, ['Age']),
        ('tension_proc', tension_pipeline, ['HasTension']),
        ('heartdisease_proc', heartdisease_pipeline, ['AnyHeartDisease']),
        ('nevermarried_proc', nevermarried_pipeline, ['NeverMarried']),
        ('occupation_proc', occupation_pipeline, ['Occupation']),
        ('livesin_proc', livesin_pipeline, ['LivesIn']),
        ('glucose_proc', glucose_pipeline, ['GlucoseLevel']),
        ('bmi_proc', bmi_pipeline, ['BMI']),
        ('smoking_proc', smokingstatus_pipeline, ['SmokingStatus'])
    ],
    remainder='drop', # Drop any columns not explicitly transformed
    verbose_feature_names_out=False # Ensures simpler column names for transformed output
)

# Fit the preprocessor on the training data and transform both training and test data
X_train_transformed = preprocessor.fit_transform(X_train_processed)
X_test_transformed = preprocessor.transform(X_test_processed)

# Print shapes to verify
print(f"Shape of X_train_transformed: {X_train_transformed.shape}")
print(f"Shape of X_test_transformed: {X_test_transformed.shape}")

# Verify feature names (for debugging/understanding, not strictly required by the prompt but good practice)
# This helps confirm the output order matches the hint.
# feature_names_out = preprocessor.get_feature_names_out()
# print("\nTransformed feature names (order check):\n", feature_names_out)
```

## Q.10 [NAT][6 marks] If you eliminate 1 feature with recursive feature elimination, which feature will be eliminated?
- Type the index of the eliminated feature (index starts from 0).
- Use `LogisticRegression` model with random state as 1729 and rest of the  parameters with default values, as an estimator.
-  Use processed training data.

V1: 15

V2: 6

V3: 6

No range needed for this question.

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

# Instantiate Logistic Regression model with random state 1729 and default parameters
estimator = LogisticRegression(random_state=1729, solver='liblinear') # 'liblinear' is often a good choice for default parameters

# Instantiate RFE with the estimator and specify to eliminate 1 feature.
# If n_features_to_select is not specified, it selects half the features.
# To find the 'first eliminated' feature, we can either set n_features_to_select = total_features - 1
# and then find the feature with rank 1, or set n_features_to_select=1 and find the feature with highest rank (meaning it was eliminated first).
# Let's set n_features_to_select to total_features - 1, and find the feature that was ranked as '1' for elimination.

# Number of features in the transformed training data
num_features = X_train_transformed.shape[1]

rfe_selector = RFE(estimator=estimator, n_features_to_select=num_features - 1, step=1, verbose=0)
rfe_selector.fit(X_train_transformed, y_train)

# The 'ranking_' attribute gives the rank of each feature (1 and up).
# Features not selected are ranked according to their elimination order (rank 1 is the first to be eliminated).
# The question asks for the feature that will be eliminated (singular), implying the *first* one eliminated.
# In RFE's ranking, '1' means it was the first one to be eliminated when selecting all but one.

eliminated_feature_index = np.where(rfe_selector.ranking_ == 1)[0][0]

print(f"The index of the first eliminated feature is: {eliminated_feature_index}")
```

###Q. 11 [MCQ][8 Marks] Out of DecisionTreeClassifier, KNeighborsClassifier and LogisticRegression, which one performs the best when used as base estimator in BaggingClassifier on the test data in terms of accuracy score when 20 base estimators are used ?

(Use random state 1729 for BaggingClassifier, DecisionTreeClassifier and LogisticRegression)

The metric for best performance will be the lowest 'absolute' difference in the train and test score.

A. DecisionTreeClassifier

B. KNeighborsClassifier

C. LogisticRegression

Answer:

V1: C

V2: C

V3: C



```python
import pandas as pd
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# --- Re-establish data splits (as y_model in kernel state is NaN) ---
# Assuming df_model is loaded from 'ModelBuilding_V1.csv' in a previous cell
# If df_model is not in the kernel state, this cell would need to load it.
# For safety, let's assume df_model was loaded correctly.

# The last column is the target column. Let's separate features (X_model) and target (y_model).
# Re-loading df_model and splitting to ensure correct y_model values
# This code block needs to be self-contained for the variable definitions.
df_model = pd.read_csv('ModelBuilding_V1.csv')
X_model = df_model.iloc[:, :-1]
y_model = df_model.iloc[:, -1]

# The following line caused the error because 'HeartAttack' is already 0/1 integers.
# y_model = y_model.map({'Yes': 1, 'No': 0})
# Removing this line as the data is already numerical.

# Manual split according to instructions: Last 30% for test, remaining for train, no shuffle.
total_rows = len(df_model)
train_size = int(total_rows * 0.7)

X_train_model = X_model.iloc[:train_size]
X_test_model = X_model.iloc[train_size:]
y_train_model = y_model.iloc[:train_size]
y_test_model = y_model.iloc[train_size:]
# ----------------------------------------------------------------------

results = {}
estimators = {
    "DecisionTreeClassifier": DecisionTreeClassifier(random_state=1729),
    "KNeighborsClassifier": KNeighborsClassifier(),
    "LogisticRegression": LogisticRegression(random_state=1729, solver='liblinear') # Using liblinear for consistency
}

for name, estimator in estimators.items():
    print(f"\n--- Evaluating {name} as base estimator ---")

    bagging_clf = BaggingClassifier(
        estimator=estimator,
        n_estimators=20,
        random_state=1729,
        n_jobs=-1 # Use all available cores for faster training
    )

    # Train the BaggingClassifier
    bagging_clf.fit(X_train_model, y_train_model)

    # Predict on training data
    y_train_pred = bagging_clf.predict(X_train_model)
    train_accuracy = accuracy_score(y_train_model, y_train_pred)

    # Predict on test data
    y_test_pred = bagging_clf.predict(X_test_model)
    test_accuracy = accuracy_score(y_test_model, y_test_pred)

    # Calculate absolute difference
    abs_diff = abs(train_accuracy - test_accuracy)

    results[name] = {"train_accuracy": train_accuracy, "test_accuracy": test_accuracy, "abs_diff": abs_diff}

    print(f"  Train Accuracy: {train_accuracy:.4f}")
    print(f"  Test Accuracy: {test_accuracy:.4f}")
    print(f"  Absolute Difference (Train-Test): {abs_diff:.4f}")

# Determine the best performing estimator
best_estimator = min(results, key=lambda k: results[k]['abs_diff'])
lowest_abs_diff = results[best_estimator]['abs_diff']

print(f"\n--- Conclusion ---")
print(f"The best performing estimator based on the lowest absolute difference (train-test score) is: {best_estimator}")
print(f"Lowest Absolute Difference: {lowest_abs_diff:.4f}")
```
